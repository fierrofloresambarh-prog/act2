```python
cal1=float(input("ingresa la primera calificación "))
cal2=float(input("ingresa la segunda calificación "))
cal3=float(input("ingresa la tercera calificación "))
calificación=(cal1+cal2+cal3)/3

print(f"La calificación promedio es: {calificación:.2f}")

if calificación < 7:
    print("reprobado")
elif calificación >= 9 and calificación >= 10:
    print("excelente aprobaste")
else: # This covers 7 <= calificación < 9
    print("aprobaste")
```
